```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# ---------------------------------------------------------
# Assignment Task 1: Load dataset and summarize basic stats
# ---------------------------------------------------------
# Load the Titanic dataset (using a reliable public source mirror)
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print("--- DATASET INFO (.info()) ---")
print(df.info())

print("\n--- BASIC STATISTICS (.describe()) ---")
print(df.describe())

# ---------------------------------------------------------
# Assignment Task 2 & Mini Project: Handle missing data
# ---------------------------------------------------------
print("\n--- MISSING VALUES BEFORE CLEANING ---")
print(df.isnull().sum())

# Impute 'Age' with median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Impute 'Embarked' with the most frequent value (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop 'Cabin' column due to excessive missing values, and drop unneeded ID/name text columns if necessary
df.drop(columns=['Cabin'], inplace=True, errors='ignore')

# ---------------------------------------------------------
# Assignment Task 3 & Mini Project: Encode Categorical Variables
# ---------------------------------------------------------
# 1. Label Encoding for 'Sex'
label_encoder = LabelEncoder()
df['Sex_Encoded'] = label_encoder.fit_transform(df['Sex'])

# 2. One-Hot Encoding for 'Embarked'
embarked_dummies = pd.get_dummies(df['Embarked'], prefix='Embarked', drop_first=True)
df = pd.concat([df, embarked_dummies], axis=1)

print("\n--- DATASET AFTER ENCODING & IMPUTATION ---")
print(df[['Name', 'Sex', 'Sex_Encoded', 'Embarked'] + list(embarked_dummies.columns)].head())

# ---------------------------------------------------------
# Mini Project Task: Visualize Age Distribution
# ---------------------------------------------------------
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'], kde=True, color='teal', bins=30)
plt.title('Titanic Passenger Age Distribution (Post-Imputation)')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('age_distribution.png', dpi=300, bbox_inches='tight')
plt.show()

# ---------------------------------------------------------
# Mini Project Task: Output cleaned dataset as new CSV
# ---------------------------------------------------------
output_filename = 'titanic_cleaned.csv'
df.to_csv(output_filename, index=False)
print(f"\nCleaned dataset successfully saved as '{output_filename}'!")
```
